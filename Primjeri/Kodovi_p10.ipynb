```python
# =========================================================
# Kod 10.1 - Rjesavanje diferentne jednacine rekurentnom formulom
#            y[n+2] - 0.8*y[n+1] + 0.25*y[n] = x[n]
#            x[n] = 10*(0.5)^n, y[0]=4, y[1]=2
# =========================================================
```

```python
N = 10                   # broj uzoraka
y = np.zeros(N)
y[0], y[1] = 4, 2         # pocetni uslovi

for n in range(N - 2):
    x_n = 10 * (0.5) ** n
    y[n + 2] = 0.8 * y[n + 1] - 0.25 * y[n] + x_n

sis.dijagram(np.arange(N), y, tip='uzorci',   
             naslov='Rjesenje diferentne jednacine $y[n]$')
```

```python
# =========================================================
# Kod 10.2 - Rjesavanje diferentne jednacine rekurentnom formulom
#            y[n+2] - 3*y[n+1] + 2*y[n] = x[n+1]
#            x[n] = 3^n * u[n], y[-1]=2, y[-2]=3
# =========================================================
```

```python
N      = 10
offset = 2   # Python ne podrzava negativne indekse: y[n] -> y[n+offset]

y = np.zeros(N + offset + 2)   # +2 zbog indeksa n+2+offset
y[-2 + offset] = 3             # y[-2] = 3
y[-1 + offset] = 2             # y[-1] = 2

for n in range(N):
    x_n1 = 3 ** (n + 1)        # x[n+1] = 3^(n+1)
    y[n+2 + offset] = 3 * y[n+1 + offset] \
                    - 2 * y[n   + offset] + x_n1

y_stem = y[offset:offset + N]

sis.dijagram(np.arange(N), y_stem, tip='uzorci', 
             naslov='Rjesenje diferentne jednacine $y[n]$')
```
