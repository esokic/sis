```python
import numpy as np
import matplotlib.pyplot as plt

# Modul sis (ako nije instaliran, npr. u Google Colab-u, instalira se automatski)
try:
    import sis
except ModuleNotFoundError:
    %pip install -q git+https://github.com/esokic/sis.git
    import sis

sis.FORMAT = "JUPYTER"    # prikaz grafika prilagodjen interaktivnom radu
```

```python
# kontinualni signal: jedinicna funkcija u(t)
t = np.arange(-5, 5.01, 0.01)

x = np.zeros(len(t))
x[t >= 0] = 1
```

```python
sis.dijagram(t, x, naslov='u(t)')
```

```python
fig, ax = plt.subplots(figsize=(4.5, 2.5))
ax.plot(t, x, color='darkgreen', linestyle='-')
ax.axhline(0, color='k', linewidth=0.5, linestyle='--')
ax.set_xlabel('$t$')
ax.set_title('u(t)')
plt.tight_layout()
plt.show()
```

```python
# diskretna sekvenca: kauzalna eksponencijalna sekvenca
n = np.arange(-3, 9)
x = 0.8 ** n * (n >= 0)
```

```python
sis.dijagram(n, x, tip='uzorci', naslov='x[n]')
```
