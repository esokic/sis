```python
# =========================================================
# Kod 2.1 - Kosinusni i sinusni signal
# =========================================================
```

```python
t = np.arange(-4, 4.001, 0.001)

x1 = np.cos(np.pi * t)
sis.dijagram(t, x1, naslov=r'$\cos(\pi t)$')
 
x2 = np.cos(np.pi * t)
sis.dijagram(t, x1, naslov=r'$\sin(\pi t)$')
```

```python
# =========================================================
# Kod 2.2 - Aperiodicki i kompleksni eksponencijalni signal
# =========================================================
```

```python
# proizvod aperiodickog i periodickog signala
t1 = np.arange(-4, 5.001, 0.001)
x1 = 2 * np.exp(-0.6 * t1) * np.sin(10 * np.pi * t1)

sis.dijagram(t1, x1, naslov=r'$x_{1}(t)=2 e^{-0.6t} \sin (10 \pi t)$')

# realni i imaginarni dio kompleksnog eksponencijalnog
t2 = np.arange(-3, 7.001, 0.001)
x2 = np.exp((-0.5 + 1j * 5 * np.pi) * t2)

sis.dijagram(t2, np.real(x2), 
             naslov = r'$\mathrm{Re}\{x_2(t) = e^{(-0.5+j5\pi)t}\}$')
sis.dijagram(t2, np.imag(x2), 
             naslov = r'$\mathrm{Im}\{x_2(t) = e^{(-0.5+j5\pi)t}\}$')
```

```python
# =========================================================
# Kod 2.3 - Zbir prigusenih sinusoida razlicitih perioda
# =========================================================
```

```python
t = np.arange(0, 5.001, 0.001)
x = (0.5 * np.exp(-t) * np.cos(19 * np.pi * t)
     + np.exp(-0.6 * t) * np.cos(21 * np.pi * t))

sis.dijagram(t, x, naslov='Zbir dvije prigusene sinusoide')
```

```python
# =========================================================
# Kod 2.4 - Jedinicna funkcija u(t)
# =========================================================
```

```python
t = np.arange(-5, 5.01, 0.01)

x = np.zeros(len(t))
x[t >= 0] = 1

sis.dijagram(t, x, naslov='$u(t)$', ylim=(-0.1, 1.1))
```

```python
# =========================================================
# Kod 2.5 - Signum funkcija sgn(t)
# =========================================================
```

```python
t = np.arange(-5, 5.01, 0.01)

y = np.zeros(len(t)) 
y[t > 0] = 1
y[t < 0] = -1

sis.dijagram(t, y, naslov='sgn(t)', ylim=(-1.1, 1.1))
```

```python
# =========================================================
# Kod 2.6 - Jedinicna rampa r(t)
# =========================================================
```

```python
t = np.arange(-1, 1.01, 0.01)

y = np.zeros(len(t))
y[t > 0] = t[t > 0]

sis.dijagram(t, y, naslov='r(t)')
```

```python
# =========================================================
# Kod 2.7 - Konvolucija dva pravougaona impulsa
# =========================================================
```

```python
Tmin = 0.01

t_x = np.arange(-1, 3 + Tmin, Tmin)
x = np.zeros(len(t_x))
x[(t_x >= 0) & (t_x <= 1)] = 1

sis.dijagram(t_x, x, naslov='$x(t)$', ylim=(-0.1, 1.1))

t_y = np.arange(-1, 3 + Tmin, Tmin)
y = np.zeros(len(t_y))
y[(t_y >= 0) & (t_y <= 1)] = 1

sis.dijagram(t_y, y, naslov='$y(t)$', ylim=(-0.1, 1.1))

t_z = np.arange(-2, 6 + Tmin, Tmin)
z = np.convolve(x, y) * Tmin   # konvolucija i skaliranje


sis.dijagram(t_z, z, 
             naslov=r'$z(t) = x(t) * y(t)$', 
             xlim = (-1, 3), ylim=(-0.1, 1.1)) 
```

```python
# =========================================================
# Kod 2.8 - Konvolucija dva pravougaona impulsa  
# =========================================================
```

```python
Tmin = 0.01

t_x = np.arange(-2, 2 + Tmin, Tmin)
x = np.zeros(len(t_x))
x[(t_x >= -0.5) & (t_x <= 0.5)] = 1

sis.dijagram(t_x, x, naslov='$x(t)$', ylim=(-0.1, 1.1))

t_y = np.arange(-2, 2 + Tmin, Tmin)
y = np.zeros(len(t_y))
y[(t_y >= -0.5) & (t_y <= 0.5)] = 1

sis.dijagram(t_y, y, naslov='$y(t)$', ylim=(-0.1, 1.1))

t_z = np.arange(-4, 4 + Tmin, Tmin)
z = np.convolve(x, y) * Tmin   # konvolucija i skaliranje


sis.dijagram(t_z, z, naslov=r'$z(t) = x(t) * y(t)$', 
             xlim=(-2,2), ylim=(-0.1, 1.1)) 
```
